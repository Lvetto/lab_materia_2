# Test delle interfacce in modalità dummy

```python
from lib.interface import *
import ipywidgets as widgets
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib widget
plt.close('all')
```

```python
interface = ElectrometerInterface(use_dummy=True)
```

```python
def clear_interface_buffers(interface):
    # Clear the current data
    interface.current_data = []
    interface.time_data = []
    interface.range_data = []
    interface.resistance_data = []
```

```python

if interface.keithley is not None:
    
    plt.figure()

    ts = np.arange(0, 2*np.pi, 0.1)
    vs = 10 * np.sin(ts) + 20

    plt.plot(ts, vs)

    print("Keithley is connected, updating dummy samples...")
    interface.keithley.set_dummy_samples(vs)

    clear_interface_buffers(interface)

    plt.show()
```
