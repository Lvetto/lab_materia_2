```python
ports = list_ports.comports()
devices = [port.device for port in ports]
for port in ports:
    print(port.device)
```

```python
for device in devices:
    try:
        print(f"Porta: {device}")
        bilancia = Bilancia(device)
        
        print(bilancia.send_command(bilancia.comandi["Receive film parameters"]))
        
        bilancia.close()
    
    except Exception as e:
        print(f"{e}")
```
