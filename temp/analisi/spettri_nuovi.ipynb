```python
nuovi_spettri_dir = "data/spettri/post_annealing"

available_spettri = [f for f in os.listdir(nuovi_spettri_dir) if f.endswith('.txt')]
available_spettri.sort()

print("Available spectra:")
for i, sp in enumerate(available_spettri):
    print(f"{i}: {sp}")
```

```python
ref_name = "ref_vetro_vetro_1_1.txt"
perc1_name = "perc_1.txt"
perc2_name = "perc_invertito_1.txt"

ref_data = np.loadtxt(os.path.join(nuovi_spettri_dir, ref_name))
lambdas = ref_data[:, 0]
ref = ref_data[:, 1]

perc1_data = np.loadtxt(os.path.join(nuovi_spettri_dir, perc1_name))
perc1 = perc1_data[:, 1]

perc2_data = np.loadtxt(os.path.join(nuovi_spettri_dir, perc2_name))
perc2 = perc2_data[:, 1]

perc1 -= ref
perc2 -= ref

plt.figure(figsize=(10, 6))

#plt.plot(lambdas, ref, label='Reference', color='black')

plt.plot(lambdas, perc1, label='Perc 1', color='blue')

plt.plot(lambdas, perc2, label='Perc 2', color='red')

plt.show()
```
